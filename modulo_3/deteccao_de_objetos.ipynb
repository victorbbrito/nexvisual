```python
import numpy as np
import cv2
import torch
import torchvision
import matplotlib.pyplot as plt
import ultralytics
```

```python
image = cv2.imread(r"C:\Users\victor.brito\Desktop\nexvisual\street.jpeg")

plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
```

```python
model = ultralytics.YOLO("yolo11n.pt")

results = model(image)

for i, result in enumerate(results):
    result.save(filename=f"result_{i}.jpg")

results_image = cv2.imread("result_0.jpg")
plt.imshow(cv2.cvtColor(results_image, cv2.COLOR_BGR2RGB))
```
