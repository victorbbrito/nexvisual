```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
```

```python
image = cv2.imread(r"C:\Users\Victor\Documents\nexvisual\bandeiranacionalbrasil.jpg")
```

```python
print(type(image))
```

```python
plt.imshow(image)
```

```python
gain = np.clip(image /255 * 1.001, 0, 1)
plt.imshow(gain)
```

```python
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
gain_rgb = np.clip(image_rgb /255 * 1.001, 0, 1)
plt.imshow(gain_rgb)
```

```python
crop = image_rgb[0:300, 0:300]
plt.imshow(crop)
```

```python
larger = cv2.resize(image_rgb, (int(image.shape[1] * 4), int(image.shape[0] * 4)))
plt.imshow(larger)
```
