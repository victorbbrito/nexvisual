```python
import numpy as np
```

### Criando vetores

```python
u = np.array([1,3])
v = np.array([3,1])
```

### Operações com vetores

```python
y = u - v
w = u + v

print(f"Vetor y: {y}\nVetor w:{w}")
```

```python
a = u / v
b = u * v

print(f"Vetor a:{a}\nVetor b:{b}")
```

```python
alpha = 2
escalar = alpha * u

print(f"Produto escalar de u por alpha: {escalar}")
```

```python
interno = np.dot(u, v)

print(f"Resultado produto interno: {interno}")

u3 = np.array([1,3,1])
v3 = np.array([3,1,1])

vetorial = np.cross(u3, v3)

print(f"Resultado produto vetorial: {vetorial}")
```

### Operações Básicas com Matrizes

```python
I = np.array([[1,0],[0,1]])
A = np.array([[2,1],[1,2]])
```

```python
B = A + I

print(f"Matriz A:\n{A}\nMatriz I:\n{I}\nMatriz B = A + I:\n{B}")
```

```python
B = A - I

print(f"Matriz A:\n{A}\nMatriz I:\n{I}\nMatriz B = A - I:\n{B}")
```

```python
C = A + u

print(f"Matriz A:\n{A}\nVetor u:{I}\nMatriz C = A + u:\n{C}")
```

```python
D = np.matmul(A, I)

print(f"Matriz D:\n{D}\nMatriz I:\n{I}\nMatriz D = A x I:\n{D}")
```
